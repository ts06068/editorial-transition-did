# Editorial transition DiD

Analysis belongs in `editorial_transition_did`. This notebook only configures and runs it.

Primary window: July 2023–June 2024 versus July 2024–June 2025. Analysis cutoff: July 24, 2026.

```python
# Keep analysis logic out of the notebook.
from editorial_transition_did import AnalysisPaths, run_all
```

## Configuration

Install the package in editable mode and create `.env`. Force a refresh only to discard the citation cache and query MSSQL.

```python
paths = AnalysisPaths.from_environment()
# True discards the cache and queries MSSQL.
FORCE_REFRESH_DB = False
PRIMARY_BOOTSTRAP_REPETITIONS = 2_000
ROBUSTNESS_BOOTSTRAP_REPETITIONS = 1_500
```

## Run the analysis

`run_all` writes the output tree beneath the configured directory.

```python
# One call owns the run. Do not duplicate pipeline steps here.
results = run_all(
    paths=paths,
    force_refresh_db=FORCE_REFRESH_DB,
    primary_bootstrap_repetitions=PRIMARY_BOOTSTRAP_REPETITIONS,
    robustness_bootstrap_repetitions=ROBUSTNESS_BOOTSTRAP_REPETITIONS,
)
```

## Data quality and validation

A failed invariant is a failed run.

```python
# Bad linkage makes every downstream estimate suspect.
display(results["source_qa"])
display(results["validation"])
```

## Difference-in-differences results

DiD is JACC's pre/post change minus the mean of the CIRC and EHJ changes. Months, not papers, are the bootstrap units.

```python
# Check the primary estimate against both robustness scales.
display(results["primary_did"])
display(results["robustness_raw"])
display(results["robustness_log"])
```
