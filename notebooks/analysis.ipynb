# Editorial transition difference-in-differences

This notebook is a thin, scientist-facing entry point for the reusable `editorial_transition_did` package. It reproduces descriptive summaries, the primary and robustness difference-in-differences analyses, and temporal citation prediction outputs for CIRC, EHJ, and JACC.

The primary pre period is July 2023 through June 2024, the post period is July 2024 through June 2025, and follow-up eligibility is evaluated at the fixed July 24, 2026 analysis cutoff.

```python
# Import only the public package surface; analysis logic remains in reusable modules.
from editorial_transition_did import AnalysisPaths, run_all
```

## Configuration

Install the project in editable mode and prepare `.env` before running. Set `FORCE_REFRESH_DB` only when the citation cache must be rebuilt.

```python
paths = AnalysisPaths.from_environment()
# Set True only to ignore and replace an existing local citation cache.
FORCE_REFRESH_DB = False
PRIMARY_BOOTSTRAP_REPETITIONS = 2_000
ROBUSTNESS_BOOTSTRAP_REPETITIONS = 1_500
```

## Run the analysis

The full pipeline writes its tables, figures, validation report, and manifest beneath the configured output directory.

```python
# One orchestrator call writes every table, figure, document, and validation file.
results = run_all(
    paths=paths,
    force_refresh_db=FORCE_REFRESH_DB,
    primary_bootstrap_repetitions=PRIMARY_BOOTSTRAP_REPETITIONS,
    robustness_bootstrap_repetitions=ROBUSTNESS_BOOTSTRAP_REPETITIONS,
)
```

## Data quality and validation

```python
# Inspect source linkage first, then the pipeline's explicit validation invariants.
display(results["source_qa"])
display(results["validation"])
```

## Difference-in-differences results

The raw estimand is the JACC post-minus-pre change minus the equally weighted mean of the CIRC and EHJ changes. Confidence intervals use a publication-month block bootstrap.

```python
# Compare the prespecified estimate with raw- and log-scale robustness results.
display(results["primary_did"])
display(results["robustness_raw"])
display(results["robustness_log"])
```
